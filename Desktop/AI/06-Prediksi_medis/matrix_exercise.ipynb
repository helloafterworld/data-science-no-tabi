# Aplikasi Vektor & Matriks: Prediksi Skor Risiko Medis

Notebook ini mendemonstrasikan bagaimana operasi matematika fundamental (vektor, matriks, dan dot product) digunakan dalam skenario medis sederhana menggunakan *neural network* mini.

**Skenario:**
Kita akan memprediksi skor risiko penyakit jantung seorang pasien berdasarkan 3 fitur:
1. Tekanan Darah Sistolik (misal: 140 mmHg)
2. Kadar Kolesterol (misal: 220 mg/dL)
3. Usia (misal: 55 tahun)

Kita akan menggunakan model dengan arsitektur: **3 input -> 2 neuron tersembunyi -> 1 output (skor risiko)**.

```python
import numpy as np
```

## Langkah 1 & 2: Input sebagai Vektor & Bobot Layer 1 sebagai Matriks

Pertama, kita ubah data pasien menjadi **vektor input (x)**. Kemudian, kita definisikan **matriks bobot (W1)** untuk lapisan pertama. Matriks ini berisi 'pengetahuan' yang telah dipelajari model untuk mengubah fitur medis mentah menjadi faktor risiko yang lebih abstrak (misal: 'faktor tekanan' dan 'faktor gaya hidup').

```python
# Data satu pasien kita representasikan sebagai vektor input (x)
pasien_input = np.array([140, 220, 55])

# Bobot (weights) dari input ke hidden layer kita representasikan sebagai matriks (W1)
# Dimensi: 2x3 (karena ada 2 neuron 'faktor risiko' & 3 fitur medis)
W1 = np.array([
    [0.05, 0.02, -0.1], # Bobot untuk 'faktor tekanan'
    [-0.01, 0.04, 0.08]  # Bobot untuk 'faktor gaya hidup'
])

# Menampilkan data awal
print(f"Vektor Input Pasien (x):\n{pasien_input}\n")
print(f"Matriks Bobot Layer 1 (W1):\n{W1}\n")
```

## Langkah 3: Menghitung Faktor Risiko Tersembunyi

Kita melakukan perkalian matriks (dot product) antara matriks bobot `W1` dan vektor input `pasien_input`. Hasilnya adalah sebuah vektor baru yang merepresentasikan "faktor risiko" tersembunyi yang telah dipelajari oleh model.

```python
# Operasi inti: Perkalian matriks antara W1 dan x
# NumPy menggunakan simbol `@` untuk perkalian matriks/dot product.
hidden_risk_factors = W1 @ pasien_input

print(f"--- Menghitung Faktor Risiko Tersembunyi ---")
print(f"Hasil W1 @ x = {hidden_risk_factors}")
print(f"Ini adalah vektor baru yang merepresentasikan 'faktor tekanan' dan 'faktor gaya hidup' pasien.\n")
```

## Langkah 4: Menuju Skor Risiko Akhir

Terakhir, kita gabungkan nilai-nilai faktor risiko tersembunyi ini untuk mendapatkan satu skor risiko akhir. Ini dilakukan dengan *dot product* antara vektor `hidden_risk_factors` dan vektor bobot kedua `W2`.

```python
# Bobot dari hidden layer (2 faktor) ke output (1 skor) sebagai vektor (W2)
W2 = np.array([0.6, 0.9])

# Menghitung prediksi akhir dengan dot product
final_risk_score = hidden_risk_factors @ W2

print(f"--- Menghitung Skor Risiko Akhir ---")
print(f"Bobot Layer 2 (W2): {W2}")
print(f"Hasil Hidden Factors @ W2 = {final_risk_score:.2f}\n")

print("="*40)
print(f"Prediksi Skor Risiko Penyakit Jantung: {final_risk_score:.2f}")
print("="*40)

# Interpretasi sederhana
if final_risk_score > 10:
    print("Interpretasi: Risiko tergolong TINGGI.")
elif final_risk_score > 5:
    print("Interpretasi: Risiko tergolong SEDANG.")
else:
    print("Interpretasi: Risiko tergolong RENDAH.")
```

## Kesimpulan
Seperti yang ditunjukkan, seluruh proses "penalaran" dari model AI sederhana ini, dari data medis mentah hingga kesimpulan skor risiko, hanyalah serangkaian operasi **vektor** dan **matriks** yang terdefinisi dengan baik. 

Saat model di-*training* dengan ribuan data pasien, yang terjadi adalah komputer secara sistematis menyesuaikan angka-angka di dalam matriks **W1** dan **W2** hingga prediksinya seakurat mungkin.
