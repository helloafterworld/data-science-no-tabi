```python
from scipy import sparse

#2D NumPy array woth diagonal 1 and 0 elsewhere
eye = np.eye(4)
print("NumPy array: \n{}".format(eye))
```

# Convert to a SciPy sparse matrix in CSR
Only the nonzero entries stored

```python
sparse_matrix = sparse.csr_matrix(eye)
print("SciPy sparsesparse matrix:\n{}".format(sparse_matrix))
```

```python
data  = np.ones(4)
row_indices = np.arange(4)
col_indices = np.arange(4)
eyes_coo = sparse.coo_matrix((data, (row_indices, col_indices)))
print("C00 representation:\n{}".format(eyes_coo))
```

# Measuring success

```python
# Making test shape
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(iris['data'], iris['target'], test_size=0.25)
print("X_test shape: {}".format(X_test.shape))
print("y_test shape: {}".format(y_test.shape))
```
